```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data (put the correct path to your file here)
df = pd.read_csv(r"C:\Users\charles\project2\Frameworks_assignment\data\metadata.csv", low_memory=False, encoding='utf-8')


# Basic exploration
print(df.shape)
print(df.info())
print(df.isnull().sum())

# Convert publish_time to datetime & extract year
df['publish_time'] = pd.to_datetime(df['publish_time'], errors='coerce')
df['year'] = df['publish_time'].dt.year

# Example: abstract word count
df['abstract_word_count'] = df['abstract'].astype(str).apply(lambda x: len(x.split()))

# Top 10 journals
top_journals = df['journal'].value_counts().head(10)
sns.barplot(y=top_journals.index, x=top_journals.values, palette='Greens')
plt.title('Top 10 Journals Publishing COVID-19 Research')
plt.xlabel('Count')
plt.ylabel('Journal')
plt.tight_layout()
plt.show()
```
